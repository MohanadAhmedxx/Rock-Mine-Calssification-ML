# ML project 1 for predicting whether Rock or Mine 
## Classification Model 


#### Importing important libraries

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

##### importing data as csv file saving it as DataFrame

```python
data = pd.read_csv(r"C:\Users\Hp\Desktop\ML and DL\Copy of sonar data.csv",header = None)
```

```python
print(data.shape)
data
```

```python
data[60].value_counts()
```

```python
X = data.drop(columns = 60 , axis = 1)
Y = data[60]
```

```python
print(type(Y))
print(type(X))
```

```python
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.1,stratify=Y,random_state=12)
print(x_train.shape , x_test.shape)
```

#### Activate Classification model  

```python
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_prediction = model.predict(x_train)
print(y_prediction)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_prediction,y_train))
```

```python
x_test_prediction = model.predict(x_test) 
print(accuracy_score(x_test_prediction , y_test) )
```
